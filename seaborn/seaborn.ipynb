```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
# scatter plot

sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title('total_bill vs tip')
plt.show()
```

```python
# line plot

sns.lineplot(x='size', y='total_bill', data=tips)
plt.title('size vs total_bill')
plt.show() 
```

```python
# categorial plots (Bar plot)

sns.barplot(x='day', y='total_bill', data=tips)
plt.title('Bar plot of day vs total_bill')  
plt.show()
```

```python
# box plot - mainly used for showing the outliers
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Box plot of day vs total_bill')  
plt.show()
```

```python
# violin plot - mainly used for showing the distribution of the data
sns.violinplot(x='day', y='total_bill', data=tips)
plt.title('Violin plot of day vs total_bill')
plt.show()
```

```python
# histogram - mainly used for showing the distribution of the data
sns.histplot(tips['total_bill'], bins=10, kde=True)
plt.title('Histogram of total_bill')
plt.show()
```

```python
# kde plot
sns.kdeplot(tips['total_bill'], fill=True)
plt.title('KDE plot of total_bill')
plt.show()
```

```python
# pair plot - mainly used for showing the relationship between multiple variables, should have numerical values
sns.pairplot(tips)  
```

```python
# heatmap - should have numerical values
corr = tips[[ 'total_bill', 'tip', 'size']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
```

```python
# some practice
sales_df = pd.read_csv('sales_data.csv')

# total sales by product
plt.figure(figsize=(10, 6))
sns.barplot(x='Product Category', y='Total Revenue', data=sales_df, estimator=sum)
plt.title('Total Sales by Product Category')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
# total sales by Region
plt.figure(figsize=(10, 6))
sns.barplot(x='Region', y='Total Revenue', data=sales_df, estimator=sum)
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.show()
```
