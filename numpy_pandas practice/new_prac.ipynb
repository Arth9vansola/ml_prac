```python
import numpy as np
import pandas as pd
```

```python
# Create a NumPy array of shape (5, 5) filled with random integers
array = np.random.randint(1, 21, size=(5, 5))
print("Original array:")
print(array)

# Replace all the elements in the third column with 1
array[:, 2] = 1
print("Modified array:")
print(array)
```

```python
# 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Replace the diagonal elements with 0.
array = np.random.randint(1, 17, size=(4, 4))
print("Original array:")
print(array)

np.fill_diagonal(array, 0)
print("Modified array:")  
print(array)
```

```python
# 1. Create a NumPy array of shape (6, 6) with values from 1 to 36. Extract the sub-array consisting of the 3rd to 5th rows and 2nd to 4th columns.
array = np.random.randint(1, 37, size=(6, 6))
print("Original array:")
print(array)


sub_arr1 = array[2:5, 1:4]
print("Extracted sub-array:")
print(sub_arr1)

# 2. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.

array2 = np.random.randint(1, 26, size=(5, 5))
print("Original array:")
print(array2)

border_elements = np.concatenate((array2[0, :], array2[-1, :], array2[1:-1, 0], array2[1:-1, -1]))
print("Border elements:")
print(border_elements)
```

```python
# 1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Add a new column that is the product of the first two columns.

df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)), columns=['A', 'B', 'C'])
print("Original DataFrame:") 
print(df)

df['D'] = df['A'] * df['B']
print("DataFrame with new column 'D':")
print(df)

# 2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.
df2 = pd.DataFrame(np.random.randint(1, 100, size=(4, 3)), columns=['A', 'B', 'C'])
print("Original DataFrame:") 
print(df2)

row_sum = df2.sum(axis=1)
col_sum = df2.sum(axis=0)
print("Row-wise sum:")
print(row_sum)
print("Column-wise sum:")
print(col_sum)

```

```python
# 1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)), columns=['A', 'B', 'C'])
print("Original DataFrame:") 
print(df)

df.loc[1, 'A'] = np.nan
df.loc[3, 'B'] = np.nan
df.loc[4, 'C'] = np.nan
  
print("DataFrame with NaN values:") 
print(df)

# Fill NaN values with the mean of the respective columns
df = df.fillna(df.mean())

print("DataFrame after filling NaN values:") 
print(df)
```

```python
# 2. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

df2 = pd.DataFrame(np.random.randint(1, 100, size=(6, 4)), columns=['A', 'B', 'C', 'D'])
print("Original DataFrame:")  
print(df2)

# Introduce some NaN values
df2.loc[1, 'A'] = np.nan
df2.loc[3, 'B'] = np.nan
df2.loc[4, 'C'] = np.nan
df2.loc[5, 'D'] = np.nan
print("DataFrame with NaN values:")
print(df2)

# Drop rows with any NaN values
df2 = df2.dropna()
print("DataFrame after dropping rows with NaN values:")
print(df2)
```
